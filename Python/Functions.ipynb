# Functions
Function is a block of reusable code that performs a specific task.

```python
def printName():
    print("Hello World")
PrintName()
PrintName()
PrintName()
```

```python
def legalOrNot(age):
    if age >= 18:
        print("Legal")
    else:
        print("Not Legal")
LegalOrNot(1)
LegalOrNot(19)
```

```python
def factorial(n):
    fact = 1
    if n == 0 or n == 1:
        print(f"Factorial of {n} is",1)
    else:
        for i in range(1,n + 1):
            fact *= i
        print(f"Factorial of num {n} is",fact)
Factorial(5)
Factorial(3)
Factorial(33)
```

```python
def printEven(List):
    for i in List:
        if i % 2 == 0:
            print(i)
printEven([1, 2, 3, 4, 5, 6, 7])
```

```python
def pyramidPattern(n):
    for i in range(1, n + 1):
        for j in range(i):
            print("* ",end="")
        print()
pyramidPattern(5)
pyramidPattern(3)
pyramidPattern(5)
```

```python
def Red():
    print("Red")
def Yellow():
    print("Yellow")
def Green():
    print("Green")
a = []
for i in range(0, 31):
    a.append(i)
for i in a:
    if i == a[len(a) - 1]:
        Green()
    else:
        if i % 2 == 0:
            Yellow()
        else:
            Red()
```

```python
def Calculator(operation, num1, num2):
    operation_l = str(operation).lower()
    if operation_l == "add" or operation_l == "addition" or operation_l == "+":
        print(f"Sum of ({num1} + {num2}) = {num1 + num2}")
    elif operation_l == "mul" or operation_l == "multiplication" or operation_l == "*":
        print(f"Product of ({num1} * {num2}) = {num1 * num2}")
    elif operation_l == "div" or operation_l == "division" or operation_l == "/":
        print(f"Quotient of ({num1} / {num2}) = {num1 / num2}")
    elif operation_l == "sub" or operation_l == "substraction" or operation_l == "-":
        print(f"Difference of ({num1} - {num2}) = {num1 - num2}")
    else:
        print("Operation not found!!")
Calculator("+", 1, 20)
```
